```python
import random
```

```python
class PrimarySkills():
    attack: int
    defense: int
    power: int
    knowledge: int
```

```python
class Hero_type():
  name: str
  primary_skills: PrimarySkills
  level_up_to_9: list
  level_after_9: list

  def total_skills(self):
    return self.primary_skills.attack + self.primary_skills.defense \
            + self.primary_skills.power + self.primary_skills.knowledge

  def skills_at_start(self):
    return [self.primary_skills.attack, self.primary_skills.defense, \
             self.primary_skills.power, self.primary_skills.knowledge]

  def give_class(self):
    return self.name

  def level_up(self, target_level, num_sim = 100, start_level = 1):

    skills_list = ['attack', 'defense', 'power', 'knowledge']
    list_attack = []
    list_defense = []
    list_power = []
    list_knowledge = []

    for i in range(num_sim):
      skills_sum = {'attack': self.primary_skills.attack,
                  'defense': self.primary_skills.defense,
                  'power': self.primary_skills.power,
                  'knowledge': self.primary_skills.knowledge}

      current_level = start_level
      for level_gained in range(start_level, target_level+1):
        if current_level <= 9:
          weights = self.level_up_to_9
        else:
          weights = self.level_after_9

        choice = random.choices(skills_list, weights=weights, k=1)[0]
        skills_sum[choice] += 1
        current_level += 1

      list_attack.append(skills_sum['attack'])
      list_defense.append(skills_sum['defense'])
      list_power.append(skills_sum['power'])
      list_knowledge.append(skills_sum['knowledge'])

    mean_attack = sum(list_attack) / len(list_attack)
    mean_defense = sum(list_defense) / len(list_defense)
    mean_power = sum(list_power) / len(list_power)
    mean_knowledge = sum(list_knowledge) / len(list_knowledge)
    total_mean = mean_attack + mean_defense + mean_power + mean_knowledge

    output_string = f"Attack: {mean_attack:.2f}\nDefense: {mean_defense:.2f}\nPower: {mean_power:.2f}\nKnowledge: {mean_knowledge:.2f}\n\nTotal: {total_mean:.2f}"
    print(output_string)
```

```python
# Redefine the subclass to inherit the updated Hero_type methods
class Beastmaster(Hero_type):
  def __init__(self):
    self.name = "Beastmaster"
    self.primary_skills = PrimarySkills()
    self.primary_skills.attack = 0
    self.primary_skills.defense = 4
    self.primary_skills.power = 1
    self.primary_skills.knowledge = 1

  level_up_to_9 = [0.30, 0.50, 0.10, 0.10]
  level_after_9 = [0.30, 0.30, 0.20, 0.20]
```

```python
class Chieftain(Hero_type):
  def __init__(self):
    self.name = "Chieftain"
    self.primary_skills = PrimarySkills()
    self.primary_skills.attack = 0
    self.primary_skills.defense = 4
    self.primary_skills.power = 1
    self.primary_skills.knowledge = 1

  level_up_to_9 = [0.35, 0.55, 0.05, 0.05]
  level_after_9 = [0.30, 0.30, 0.20, 0.20]
```

```python
alora = Chieftain()
alora.level_up(19)
```
